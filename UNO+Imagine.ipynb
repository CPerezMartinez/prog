```python
import pandas as pd

df=pd.read_csv("temperatura coso - Full 1 (1).csv")
df
```

```python
#Màxim i mínim aula a-14
max_u = df['Temperatura a-14'].max()
print("El màxim de tempertura a l'aula a-14 és", max_u, "ºC")
min_u = df['Temperatura a-14'].min()
print("El mínim de tempertura a l'aula a-14 és", min_u, "ºC")
```

```python
#Màxim i mínim al passadís
max_u = df['Temperatura passadís'].max()
print("El màxim de tempertura al passadís és", max_u, "ºC")
min_u = df['Temperatura passadís'].min()
print("El mínim de tempertura al passaddís és", min_u, "ºC")
```

```python
#Màxim i mínim al pati
max_u = df['Temperatura pati'].max()
print("El màxim de tempertura al pati és", max_u, "ºC")
min_u = df['Temperatura pati'].min()
print("El mínim de tempertura al pati és", min_u, "ºC")
```

```python
#Màxim i mínim al gimnàs
max_u = df['Temperatura gimnàs'].max()
print("El màxim de tempertura al gimnàs és", max_u, "ºC")
min_u = df['Temperatura gimnàs'].min()
print("El mínim de tempertura al gimnàs és", min_u, "ºC")
```

```python
#Màxim i mínim aula a-23
max_u = df['Temperatura a-23'].max()
print("El màxim de tempertura a l'aula a-23 és", max_u, "ºC")
min_u = df['Temperatura a-23'].min()
print("El mínim de tempertura a l'aula a-23 és", min_u, "ºC")
```

```python
#Mitjana ponderada
suma_u = df["Temperatura a-14"].sum()
promig_u = (suma_u/179)
suma_dos = df["Temperatura passadís"].sum()
promig_dos = (suma_dos/179)
suma_tres = df["Temperatura pati"].sum()
promig_tres = (suma_tres/179)
suma_quatre = df["Temperatura gimnàs"].sum()
promig_quatre = (suma_dos/179)
suma_cinc = df["Temperatura a-23"].sum()
promig_cinc = (suma_tres/179)

suma_total = (int(promig_u+promig_dos+promig_tres+promig_quatre+promig_cinc))
promig = (suma_total/5)
promig_total = (round(promig,2))
print("El promig de la temperatura és", promig_total, "ºC")
```
